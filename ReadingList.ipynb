## Gaps noted

- Kathryn (2026-10-20): no satisfying account of LLMs from a probabilistic or likelihood-based view for statisticians. Also missing: what token probabilities, model confidence, calibration and stochastic sampling mean, and how they relate to statistical inference. Ji et al. touch on these but not in depth.
- Farouk (2026-10-20): a possible review paper, "Large Language Models: a Review for Statisticians" or "Understanding the Nuts and Bolts of Large Language Models", if a useful angle exists.
